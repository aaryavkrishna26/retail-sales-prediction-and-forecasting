# 02. Preprocessing & Feature Engineering
## Retail Sales Predictive Analytics & Time Series Forecasting

### Objectives:
1. Merge store metadata, interpolated oil prices, and active holiday events.
2. Extract calendar signals (day of week, month, year, bi-weekly payday indicators).
3. Build lag and rolling window features strictly respecting test horizon boundaries (>= 16-day lag).
4. Formulate the surge classification target (`is_high_sales`) for logistic modeling.
5. Execute temporal holdout split (train up to 2017-07-31, validation 2017-08-01 to 2017-08-15).

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
import numpy as np
from src.data_loader import load_raw_datasets
from src.preprocess import (
    merge_external_features,
    add_lag_and_rolling_features,
    create_surge_classification_target,
    temporal_train_val_split
)
from src.config import PROCESSED_DATA_DIR
```

```python
raw_data = load_raw_datasets()
if "train" in raw_data:
    merged_df = merge_external_features(
        sales_df=raw_data["train"],
        stores_df=raw_data["stores"],
        oil_df=raw_data["oil"],
        holidays_df=raw_data["holidays"]
    )
    engineered_df = add_lag_and_rolling_features(merged_df)
    engineered_df = create_surge_classification_target(engineered_df)
    train_df, val_df = temporal_train_val_split(engineered_df)
    print(f"Engineered dataset ready: Train {train_df.shape}, Val {val_df.shape}")
```
